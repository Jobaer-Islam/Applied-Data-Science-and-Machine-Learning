```python
clt_pop = sp.expon.rvs(scale = 10, size =10000)
clt_pop
```

```python
clt_mean  = np.mean(clt_pop)
clt_std = np.std(clt_pop)
```

```python
clt_mean
```

```python
clt_std
```

```python
n_samples = 10000
```

```python
sample_size = 1000
sample_mean = []

for _ in range(sample_size):
    sample = np.random.choice(clt_pop, size = sample_size)
    sample_mean.append(np.mean(sample))

len(sample_mean)

plt.figure(figsize = (12, 5))
plt.subplot(1, 2,1)
plt.hist(clt_pop, bins=50, density=True, color="blue", alpha = 0.7, label = "Sample Mean")
plt.title("Jobaer Islam")
plt.xlabel("value data")
plt.ylabel("Density")



plt.subplot(1, 2,2)
plt.hist(sample_mean, bins=50, density=True, color="lightgreen", alpha = 0.7)
x_axis = np.linspace(min(sample_mean) , max(sample_mean), 100)

std_error  = clt_std / np.sqrt(sample_size)
plt.plot(x_axis, sp.norm.pdf(x_axis, clt_mean , std_error), 'r-', lw = 2, label = "Bell Curve")
plt.title("Jobaer Islam")
plt.xlabel("value data")
plt.ylabel("Density")

plt.show()
```

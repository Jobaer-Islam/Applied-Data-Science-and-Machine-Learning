```python
import numpy as np
from scipy import stats
import statsmodels.api as sm
import pandas as pd
import matplotlib.pyplot as plt
np.random.seed(42)
```

## 1.1. Handling Uncertainty and Variability

```python
spend_data = np.array([15, 120])
data_mean = np.mean(spend_data)
data_std = np.std(spend_data, ddof=1)
print(f"User spend data: {spend_data}")
print(f"Average (Mean) spend: ${data_mean:.2f}")
print(f"Sample Standard Deviation: ${data_std:.2f}")
print("\n")
```

## 1.2. Moving from Data Summaries to Data-Driven Decisions

### 1.2: A/B Testing

```python
contingency_table = np.array([
    [80, 920],
    [95, 905]
])

chi2_stat, p_value, dof, expected_freqs = stats.chi2_contingency(contingency_table)

print("Contingency Table (Clicks vs. No Clicks):")
print(contingency_table)
print(f"\nChi-Squared Statistic: {chi2_stat:.4f}")
print(f"P-Value: {p_value:.4f}")

alpha = 0.05
print(f"\nSignificance Level (alpha): {alpha}")

if p_value < alpha:
    print(f"Result (p={p_value:.4f}): REJECT the Null Hypothesis.")
    print("The difference between the buttons is statistically significant.")
else:
    print(f"Result (p={p_value:.4f}): FAIL to Reject the Null Hypothesis.")
    print("The difference is not statistically significant; it could be due to random chance.")

# Note: The lecture's p-value of 3% (0.03) was a hypothetical example.
# Our actual data (80 vs 95) gives a p-value of ~0.2679, which is
# not significant at the 0.05 level.
print("\n")
```

## 1.3. Foundation for Inference and Model Evaluation

```python
n_samples = 1000
true_sqft_weight = 150
true_intercept = 75000
X_sqft = np.random.normal(loc=2000, scale=300, size=n_samples)
Y_noise = np.random.normal(loc=0, scale=40000, size=n_samples)
Y_price = (X_sqft * true_sqft_weight) + true_intercept + Y_noise
X_with_const = sm.add_constant(X_sqft)
model = sm.OLS(Y_price, X_with_const)
results = model.fit()
```

```python
results
```

### 1.3a: Confidence Intervals

```python
print(f"Simulated {n_samples} houses with TRUE SqFt weight = {true_sqft_weight}")
```

```python
ci = results.conf_int(alpha=0.05)
coeff_df = pd.DataFrame(ci, columns=['Lower 95%', 'Upper 95%'])
coeff_df['Point Estimate'] = results.params

coeff_df.index = ['Intercept', 'SqFt_Weight']
print("\nLinear Regression Results:")
print(coeff_df)
print("\nInterpretation: Our model estimated the SqFt weight (Point Estimate)")
print("and calculated a 95% confidence interval for it.")
print(f"The true value of {true_sqft_weight} is within this interval.")
print("\n")
```

### 1.3b: Paired T-Test

```python
n_test_houses = 200
errors_A = np.random.normal(loc=28000, scale=5000, size=n_test_houses)
errors_B = errors_A - np.random.normal(loc=3000, scale=1000, size=n_test_houses)

print(f"Model A average error: ${np.mean(errors_A):.2f}")
print(f"Model B average error: ${np.mean(errors_B):.2f}")

t_stat, p_value_ttest = stats.ttest_rel(errors_A, errors_B, alternative='greater')

print(f"\nPaired T-Test Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value_ttest:.4e}") # Use scientific notation for very small p-values

if p_value_ttest < alpha:
    print(f"Result (p={p_value_ttest:.4e}): REJECT the Null Hypothesis.")
    print("Model B's lower error is a statistically significant improvement.")
else:
    print(f"Result (p={p_value_ttest:.4e}): FAIL to Reject the Null Hypothesis.")
    print("The difference in model performance is not significant.")
```

## 2.1. Measures of Central Tendency

```python
data = [150, 160, 160, 180, 200, 210, 800]
s = pd.Series(data)

mean_val = s.mean()
print(f"Data: {data}")
print(f"Mean:   {mean_val:.2f}  (Matches lecture: 265.7)")

median_val = s.median()
print(f"Median: {median_val}  (Matches lecture: 180)")

mode_val = s.mode()[0]
print(f"Mode:   {mode_val}  (Matches lecture: 160)")
print("\n")

print("--- Demonstration: Sensitivity to Outliers ---")
data_no_outlier = [150, 160, 160, 180, 200, 210]
s_no_outlier = pd.Series(data_no_outlier)

mean_no_outlier = s_no_outlier.mean()
median_no_outlier = s_no_outlier.median()

print(f"Original Mean:     {mean_val:.2f}   | New Mean:     {mean_no_outlier:.2f}  (Changes a lot!)")
print(f"Original Median:   {median_val}    | New Median:   {median_no_outlier}   (Very stable/robust)")
print("\n")
```

```python
var_val = s.var()
print(f"Variance (s^2): {var_val:.2f}")
std_val = s.std()
print(f"Standard Dev (s): {std_val:.2f}")

range_val = s.max() - s.min()
print(f"Range: {range_val}  (Matches lecture: 800 - 150 = 650)")

Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)
IQR = Q3 - Q1
print(f"Q1 (25th percentile): {Q1}")
print(f"Q3 (75th percentile): {Q3}")
print(f"IQR (Q3 - Q1): {IQR}  (Matches lecture: 210 - 160 = 50)")
print("\n")

print("--- Data Science Shortcut: .describe() ---")
print(s.describe())
print("\n")
```

## 2.3. Correlation vs. Causation Example

```python
np.random.seed(42)
num_days = 100
hot_weather = np.random.rand(num_days) * 20 + 15
ice_cream_sales = (hot_weather - 15) * 5 + np.random.randn(num_days) * 5 + 20
ice_cream_sales = np.clip(ice_cream_sales, 0, 1000)
shark_attacks = (hot_weather - 20) * 0.5 + np.random.randn(num_days) * 0.5
shark_attacks = np.clip(shark_attacks, 0, 10)

spurious_corr = np.corrcoef(ice_cream_sales, shark_attacks)[0, 1]
print(f"Spurious Correlation (Ice Cream vs. Sharks): {spurious_corr:.4f}")

real_corr_1 = np.corrcoef(ice_cream_sales, hot_weather)[0, 1]
real_corr_2 = np.corrcoef(shark_attacks, hot_weather)[0, 1]
print(f"Real Correlation (Ice Cream vs. Weather):   {real_corr_1:.4f}")
print(f"Real Correlation (Sharks vs. Weather):      {real_corr_2:.4f}")
```

```python
import numpy as np
Omega_die = {1, 2, 3, 4, 5, 6}
A_even = {2, 4, 6}
B_gt4 = {5, 6}
P_A = len(A_even) / len(Omega_die)
P_B = len(B_gt4) / len(Omega_die)

print(f"Sample Space (Omega): {Omega_die}")
print(f"Event A (Even): {A_even}")
print(f"Event B (> 4): {B_gt4}")
print(f"P(A) = {len(A_even)} / {len(Omega_die)} = {P_A:.4f}  (Matches lecture: 0.5)")
print(f"P(B) = {len(B_gt4)} / {len(Omega_die)} = {P_B:.4f}  (Matches lecture: 0.33)")
print("\n")
```

## 3.2. Conditional Probability

```python
A_in_B = A_even.intersection(B_gt4)
P_A_given_B_intuitive = len(A_in_B) / len(B_gt4)

print("Method 1 (Intuitive):")
print(f"  New Sample Space (B): {B_gt4}")
print(f"  Event A within B: {A_in_B}")
print(f"  P(A|B) = {len(A_in_B)} / {len(B_gt4)} = {P_A_given_B_intuitive:.4f}  (Matches lecture: 0.5)")
A_intersect_B = A_even.intersection(B_gt4)
P_A_intersect_B = len(A_intersect_B) / len(Omega_die)
P_A_given_B_formal = P_A_intersect_B / P_B
```

```python
print("\nMethod 2 (Formal Formula):")
print(f"  P(A and B) = {len(A_intersect_B)} / {len(Omega_die)} = {P_A_intersect_B:.4f}")
print(f"  P(B) = {P_B:.4f}")
print(f"  P(A|B) = {P_A_intersect_B:.4f} / {P_B:.4f} = {P_A_given_B_formal:.4f}  (Matches lecture: 0.5)")
print("\n")
```

```python
print("--- The Chain Rule ---")
# P(A and B) = P(A | B) * P(B)
P_A_intersect_B_calc = P_A_given_B_formal * P_B
print(f"P(A and B) = {P_A_given_B_formal:.4f} * {P_B:.4f} = {P_A_intersect_B_calc:.4f}")
print(f"Matches our earlier P(A and B): {P_A_intersect_B_calc == P_A_intersect_B}")
print("\n")
```

```python
print(f"Is P(A|B) == P(A)?")
print(f"  P(A|B) = {P_A_given_B_formal:.4f}")
print(f"  P(A)   = {P_A:.4f}")
# Note: They are equal (0.5 == 0.5), so they are independent!
print(f"Result: {P_A_given_B_formal == P_A} (They are independent)")
```

```python
parent_population = stats.expon.rvs(scale=10, size=100000)
pop_mean = np.mean(parent_population)
pop_std = np.std(parent_population)
plt.figure(3, figsize=(10, 5))
plt.hist(parent_population, bins=100, density=True, alpha=0.7, label='Parent Population')
plt.title('Parent Population (Skewed)', fontsize=16)
plt.legend()
plt.show()
```

```python
num_samples = 10000
sample_size = 30
sample_means = []

for i in range(num_samples):
    sample = np.random.choice(parent_population, size=sample_size)
    sample_mean = np.mean(sample)
    sample_means.append(sample_mean)


plt.figure(4, figsize=(10, 5))
plt.hist(sample_means, bins=70, density=True, alpha=0.7, label='Sampling Distribution')
plt.title('Distribution of Sample Means (n=30)', fontsize=16)
```

```python
clt_mean = np.mean(sample_means)
clt_std_theory = pop_std / np.sqrt(sample_size)
clt_std_observed = np.std(sample_means)

print(f"Parent Mean (μ):    {pop_mean:.4f}")
print(f"CLT Mean (Observed): {clt_mean:.4f}  (They are almost identical!)")
print("")
print(f"Parent STD (σ):     {pop_std:.4f}")
print(f"CLT STD (Theory):   {clt_std_theory:.4f}  (σ / sqrt(n))")
print(f"CLT STD (Observed): {clt_std_observed:.4f}  (They are almost identical!)")

x = np.linspace(min(sample_means), max(sample_means), 200)
pdf = stats.norm.pdf(x, loc=clt_mean, scale=clt_std_observed)
plt.plot(x, pdf, 'r-', linewidth=2, label='Normal Distribution (CLT)')
plt.legend()
plt.show()
```

```python
plot_orgin =np.array([0, 0])
plot_orgin
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize = (10, 8))
fig, ax = plt.subplots()

ax.set_title("Lecture 05: Math for Data Science I (Linear Algebr")
ax.grid(True)
ax.set_axisbelow(True)


ax.quiver(*plot_orgin, *u, angles = "xy", scale_units = "xy", scale = 1, color = "blue", label  ="U" )
ax.quiver(*u, *v, angles = "xy",  scale_units = "xy",  scale = 1, color = "red",  label  = "V")
ax.quiver(*plot_orgin,*vaddu, angles = "xy",  scale_units = "xy",  scale = 1, color = "green",  label  = "U+V")
ax.plot([v[0], vaddu[0]], [v[1], vaddu[1]], 'r:')
ax.plot([plot_orgin[0], v[0]],[plot_orgin[1], v[1]], 'r:')


ax.set_xticks(range(0,  6))
ax.set_yticks(range(0, 6))
ax.set_xlim(0,  5)
ax.set_ylim(0, 5)
ax.set_aspect("equal")
ax.legend()
plt.show()
```
